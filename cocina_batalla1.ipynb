```python
class Cuchillo:
    def __init__(self):
        self.filo = 100

    def cortar(self, ingrediente):
        if self.filo <= 0:
            print(" El cuchillo ha perdido su poder. Requiere ejecutar el método Afilar.")
            return

        print(f" Cortando {ingrediente.nombre}...")
        ingrediente.vida -= 50
        self.filo -= 5

        if ingrediente.vida < 0:
            ingrediente.vida = 0

        print(f"Vida de {ingrediente.nombre}: {ingrediente.vida}%")
        print(f"Filo del cuchillo: {self.filo}%")

        if ingrediente.vida == 0:
            ingrediente.morir()

    def afilar(self):
        self.filo = 100
        print(" El cuchillo ha sido afilado. Filo restaurado al 100%.")

```

```python

class Ingrediente:
    def __init__(self, nombre):
        self.nombre = nombre
        self.vida = 100
        self.procesado = False

    def morir(self):
        self.procesado = True
        print(f" {self.nombre} está completamente picado.")
```

```python
class Pasta:
    def __init__(self):
        self.coccion = 0

    def subir_nivel(self):
        self.coccion += 20
        if self.coccion > 100:
            self.coccion = 100
        print(f" Cocción de la pasta: {self.coccion}%")
```

```python
class Olla:
    def __init__(self):
        self.contenido = []

    def agregar(self, item):
        self.contenido.append(item)
        print(f" {item} agregado a la olla.")
```

```python
inventario = ["Olla", "Cebolla", "Tomate", "Pastas", "Agua"]

requeridos = ["Olla", "Cebolla", "Tomate", "Pastas", "Agua"]

print(" Verificando inventario...")

if all(item in inventario for item in requeridos):
    print(" Todos los objetos requeridos están disponibles.")
else:
    print(" Faltan ingredientes. No puedes comenzar.")
    exit()
```

```python
cuchillo = Cuchillo()
cebolla = Ingrediente("Cebolla")
tomate = Ingrediente("Tomate")
pasta = Pasta()
olla = Olla()
```

```python
def procesar_ingrediente(ingrediente):
    while not ingrediente.procesado:
        print("\n1. Cortar")
        print("2. Afilar")
        opcion = input("Elige una opción: ")

        if opcion == "1":
            cuchillo.cortar(ingrediente)
        elif opcion == "2":
            cuchillo.afilar()
        else:
            print("Opción inválida.")


print("\n Combate: El cuchillo vs. La Cebolla")
procesar_ingrediente(cebolla)

print("\n Combate: El cuchillo vs. El Tomate")
procesar_ingrediente(tomate)
```

```python
print("\n Ensamblando ingredientes en la olla...")
olla.agregar("Cebolla picada")
olla.agregar("Tomate picado")
olla.agregar("Pastas")
olla.agregar("Agua")
```

```python
s
print("\n Comienza el proceso de cocción...")

while pasta.coccion < 100:
    decision = input("¿Deseas seguir hirviendo? (s/n): ").lower()

    if decision == "s":
        pasta.subir_nivel()
    else:
        print(" La cocción se detuvo antes de tiempo.")
        break


if pasta.coccion == 100:
    print("\n ¡Combate culinario finalizado!")
else:
    print("\n La pasta no alcanzó su punto perfecto.")
```
